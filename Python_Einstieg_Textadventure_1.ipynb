## 4. Rechnen mit Zahlen

Variablen können nicht nur Text, sondern auch **Zahlen** speichern.
Zahlen schreibst du **ohne** Anführungszeichen – sonst denkt Python, es ist Text!

```python
gold = 100        # Zahl
name = "Aria"     # Text (mit Anführungszeichen)
```

Mit Zahlen kannst du rechnen:

| Zeichen | Bedeutung | Beispiel |
|---|---|---|
| `+` | Plus | `gold + 50` |
| `-` | Minus | `gold - 30` |
| `*` | Mal | `gold * 2` |
| `+=` | Erhöhen um | `gold += 50` → gold wird 50 größer |
| `-=` | Verringern um | `gold -= 30` → gold wird 30 kleiner |

````python
# Demo
gold = 0
leben = 5

print("Start – Gold:", gold, "| Leben:", leben)

gold += 100          # Schatztruhe gefunden
print("Schatztruhe gefunden! Gold:", gold)

gold -= 30           # Trank gekauft
leben += 1
print("Trank gekauft. Gold:", gold, "| Leben:", leben)

gold -= 20           # Brücken-Zoll bezahlt
print("Zoll bezahlt. Gold:", gold)
````

## 7. Alles zusammen: Dein eigenes Textadventure!

Jetzt kombinierst du alles. Hier ist ein Beispiel als Vorlage – danach bist du dran!

**So funktioniert ein Textadventure:**
- Jede Szene ist eine Funktion
- In jeder Szene gibt es eine Auswahl per `input()`
- Je nach Eingabe wird die nächste Szene (Funktion) aufgerufen

````python
# Beispiel-Textadventure (zum Ausprobieren)

def start():
    print("\nDu stehst an einer Wegkreuzung.")
    print("1: Gehe nach links (zum Turm)")
    print("2: Gehe nach rechts (zum Dorf)")
    wahl = input("Was tust du? ")

    if wahl == "1":
        turm()
    elif wahl == "2":
        dorf()
    else:
        print("Ungültige Eingabe!")
        start()

def turm():
    print("\nDu erreichst einen alten Turm. Die Tür steht offen.")
    print("1: Betritt den Turm")
    print("2: Kehre um")
    wahl = input("Was tust du? ")

    if wahl == "1":
        print("Im Turm findest du einen Zauberstab. Du hast gewonnen!")
    elif wahl == "2":
        start()
    else:
        print("Ungültige Eingabe!")
        turm()

def dorf():
    print("\nIm Dorf ist es ruhig. Ein Händler spricht dich an.")
    print("1: Kaufe ein Schwert (kostet alles dein Gold)")
    print("2: Ignoriere ihn und gehe weiter")
    wahl = input("Was tust du? ")

    if wahl == "1":
        print("Mit dem Schwert wirst du zum größten Helden des Landes!")
    elif wahl == "2":
        print("Du gehst weiter, doch ohne Ausrüstung. Ende.")
    else:
        print("Ungültige Eingabe!")
        dorf()

# Spiel starten!
print("=" * 40)
print("   WILLKOMMEN ZUM TEXTADVENTURE")
print("=" * 40)
start()
````

## 8. Schleifen mit `while`

Mit `while` wiederholt Python einen Code-Block, solange eine Bedingung wahr ist.

```python
while Bedingung:
    # wird immer wieder ausgeführt, solange Bedingung wahr ist
```

Das ist besonders nützlich, wenn du etwas wiederholen willst – zum Beispiel solange der Held noch am Leben ist, oder bis der Spieler eine gültige Eingabe macht.

> **Wichtig:** Die Bedingung muss irgendwann **falsch** werden – sonst läuft die Schleife ewig!

````python
# Demo
leben = 3

print("=== Ein Gegner greift an! ===")
while leben > 0:
    print("Du hast noch", leben, "Leben.")
    wahl = input("Was tust du? (1=Kämpfen / 2=Fliehen) ")

    if wahl == "1":
        leben -= 1  # Du triffst ihn, aber er trifft dich auch
        print("Du schlägst zu! Aber du wirst auch getroffen.")
    elif wahl == "2":
        print("Du fliehst in Sicherheit!")
        leben = 0   # Schleife beenden
    else:
        print("Das verstehe ich nicht. Versuch es nochmal.")

print("Der Kampf ist vorbei.")
````

## 9. Listen als Inventar

Eine **Liste** speichert mehrere Werte auf einmal – wie ein Rucksack für dein Abenteuer!

```python
inventar = []                    # leere Liste erstellen
inventar.append("Schwert")       # Gegenstand hinzufügen
inventar.remove("Schwert")       # Gegenstand entfernen
"Schwert" in inventar            # prüfen ob vorhanden → True oder False
print(inventar)                  # Liste anzeigen
```

Listen erkennst du an den **eckigen Klammern** `[ ]`. Mehrere Einträge werden durch Kommas getrennt: `["Fackel", "Schwert", "Schlüssel"]`.

````python
# Demo
inventar = []

print("Dein Inventar:", inventar)

inventar.append("Fackel")
inventar.append("Schlüssel")
print("Nach dem Höhleneingang:", inventar)

# Mit dem Schlüssel eine Tür öffnen
if "Schlüssel" in inventar:
    print("Du öffnest die Tür mit dem Schlüssel!")
    inventar.remove("Schlüssel")
else:
    print("Du hast keinen Schlüssel.")

print("Inventar am Ende:", inventar)
````

## 11. f-Strings: Variablen im Text

Bisher hast du Variablen mit Komma in `print()` eingebaut:
```python
print("Gold:", gold, "| Leben:", leben)
```

Mit einem **f-String** geht das viel übersichtlicher – du schreibst die Variable
einfach direkt in den Text, in geschweiften Klammern `{ }`:
```python
print(f"Gold: {gold} | Leben: {leben}")
```

Das kleine **`f`** vor dem Anführungszeichen macht den String zum f-String.
Alles in `{ }` wird automatisch durch den Wert der Variable ersetzt.

````python
# Demo
name = "Aria"
gold = 120
leben = 4
ort = "Wald"

# Ohne f-String (unübersichtlich)
print("Held:", name, "| Ort:", ort, "| Gold:", gold, "| Leben:", leben)

# Mit f-String (viel lesbarer)
print(f"Held: {name} | Ort: {ort} | Gold: {gold} | Leben: {leben}")

# f-Strings funktionieren auch mit Rechnen direkt im Text
print(f"Mit einem Trank hättest du {leben + 2} Leben.")
````

## 12. Programmstruktur

Je größer dein Textadventure wird, desto wichtiger ist eine klare Struktur.
Halte dich immer an dieses Muster:

```python
# 1. Funktionsdefinitionen
def szene_wald():
    ...

def main():
    ...

# 2. Programstart
# 3. Globale Variablen und Listen
gold = 0
inventar = []

# 4. Hauptfunktion aufrufen
main()
```

> **Warum `main()` ganz unten?** Python liest den Code von oben nach unten.
> Eine Funktion muss erst *definiert* sein, bevor sie *aufgerufen* werden kann.
> Wenn du `main()` ganz unten schreibst, sind garantiert alle Funktionen bekannt.

````python
# Demo

# 1. Funktionsdefinitionen
def begruessung():
    print(f"=== Willkommen, {held_name}! ===")

def wald():
    print(f"{held_name} betritt den Wald. Inventar: {inventar}")
    inventar.append("Fackel")
    print(f"Du findest eine Fackel! Inventar: {inventar}")

def main():
    begruessung()
    wald()
    print("Abenteuer beendet.")

# 2. Programstart
# 3. Globale Variablen und Listen
held_name = "Aria"
inventar = []

# 4. Hauptfunktion aufrufen
main()
````

## 13. Namespaces: Lokale und globale Variablen

In Python hat jede Variable einen **Geltungsbereich** (englisch: *namespace*) –
also einen Bereich, in dem sie bekannt ist.

| | Wo definiert? | Wo sichtbar? |
|---|---|---|
| **Lokale Variable** | Innerhalb einer Funktion | Nur in dieser Funktion |
| **Globale Variable** | Außerhalb aller Funktionen | Überall im Programm |

**Lesen** einer globalen Variable aus einer Funktion heraus klappt problemlos.
**Schreiben** (also verändern mit `+=`, `-=`, `=`) ist das Problem:

```python
gold = 100

def verdiene_gold():
    gold += 50    # ✗ NameError! Python sieht das += und denkt,
                  #   gold soll eine NEUE lokale Variable sein –
                  #   die aber nie einen Startwert bekommen hat.
```

Die Lösung ist das **`global`-Keyword**. Es sagt Python: „Diese Variable
gehört zum globalen Bereich, nicht zu diesem Funktions-Block."

```python
def verdiene_gold():
    global gold   # ✓ jetzt weiß Python: gemeint ist die globale Variable
    gold += 50
```

> **Faustregel:** Listen verändern (`.append()`, `.remove()`) geht ohne `global`,
> weil du die Liste nicht neu zuweist, sondern nur ihren Inhalt veränderst.
> Zahlen und Texte hingegen brauchst du `global`, sobald du sie mit `=` oder `+=` veränderst.

````python
# Demo

# 1. Funktionsdefinitionen
def schatz_finden():
    global gold          # Zahl verändern → global nötig
    gold += 50
    inventar.append("Karte")   # Liste verändern → kein global nötig
    print(f"Schatz! Gold: {gold} | Inventar: {inventar}")

def heiltrank_trinken():
    global leben
    leben += 2
    print(f"Heiltrank getrunken. Leben: {leben}")

def status():
    # Nur lesen → kein global nötig
    print(f"Gold: {gold} | Leben: {leben} | Inventar: {inventar}")

def main():
    status()
    schatz_finden()
    heiltrank_trinken()
    status()

# 2. Programstart
# 3. Globale Variablen und Listen
gold = 0
leben = 3
inventar = []

# 4. Hauptfunktion aufrufen
main()
````

## 15. Rückgabewerte mit `return`

Bisher haben Funktionen nur Dinge *ausgegeben* (`print`).
Mit `return` kann eine Funktion auch einen Wert *zurückgeben* –
den du dann weiterbenutzen kannst.

```python
def verdopple(zahl):
    return zahl * 2

ergebnis = verdopple(5)   # ergebnis = 10
print(ergebnis)           # 10
```

Der Unterschied:

| | Was passiert? | Wert weiternutzbar? |
|---|---|---|
| `print(x)` | Zeigt x auf dem Bildschirm | Nein |
| `return x` | Gibt x an den Aufrufer zurück | Ja |

Funktionen können auch `True` oder `False` zurückgeben –
das ist sehr nützlich für Bedingungen:
```python
def hat_genug_gold(gold, preis):
    return gold >= preis

if hat_genug_gold(80, 50):
    print("Du kannst es kaufen!")
```

````python
# Demo

def kampf_gewonnen(held_staerke, gegner_staerke):
    return held_staerke > gegner_staerke   # gibt True oder False zurück

def beute_berechnen(anzahl_gegner):
    return anzahl_gegner * 15              # gibt eine Zahl zurück

def zusammenfassung(gold, inventar):
    return f"Gold: {gold} | Inventar: {inventar}"  # gibt Text zurück


# Rückgabewerte verwenden
sieg = kampf_gewonnen(8, 5)
if sieg:
    beute = beute_berechnen(3)
    print(f"Du gewinnst den Kampf! Beute: {beute} Gold")
else:
    print("Du verlierst den Kampf.")

inventar = ["Schwert", "Fackel"]
gold = 120
print(zusammenfassung(gold, inventar))
````
